# Lecture 01: Mathematics Primer

**Matrix Basics for Machine Learning**

## Big Idea
Matrices transform space, and understanding these transformations helps us grasp how machine learning algorithms process and interpret data.

## Everyday Metaphor

> 💡 **Matrix = rubber-sheet stretching machine**: Imagine a grid drawn on a rubber sheet. When you pull the sheet in different directions, the grid lines stretch and rotate, but the relationship between points stays consistent. 
> 
> **Determinant = area scaling factor**: Tells you how much the rubber sheet stretches or shrinks areas. A determinant of 2 means areas become twice as big; 0.5 means half the size; 0 means the sheet collapses to a line.
> 
> **Eigenvector = direction that only stretches**: Certain directions on the rubber sheet, when pulled, don't change direction - they only get longer or shorter. These special directions are eigenvectors, and how much they stretch is the eigenvalue.

## Why It Matters
- In machine learning, data is often represented as matrices
- Understanding matrix operations helps us comprehend how algorithms like PCA reduce dimensionality
- Helps us understand how neural networks transform inputs through layers
- Enables us to solve systems of equations that arise in model fitting
- Without this foundation, we'd be using ML tools as black boxes

## Formulas


### Determinant of 2x2 Matrix

$$
\text{det}\begin{pmatrix} a & b \\ c & d \end{pmatrix} = ad - bc
$$

*What each symbol means:*
- **a, b, c, d**: The four entries of the 2x2 matrix
- **ad**: Product of the diagonal from top-left to bottom-right
- **bc**: Product of the diagonal from top-right to bottom-left
- **ad - bc**: The difference gives the signed area scaling factor

### Determinant of 3x3 Matrix

$$
\text{det}\begin{pmatrix} a & b & c \\ d & e & f \\ g & h & i \end{pmatrix} = a(ei - fh) - b(di - fg) + c(dh - eg)
$$

*What each symbol means:*
- **a, b, c, d, e, f, g, h, i**: The nine entries of the 3x3 matrix
- **ei - fh**: Determinant of the submatrix obtained by removing row 1 and column 1
- **di - fg**: Determinant of the submatrix obtained by removing row 1 and column 2
- **dh - eg**: Determinant of the submatrix obtained by removing row 1 and column 3
- The signs follow a + - + pattern for the first row expansion

### Trace of a Matrix

$$
\text{tr}(A) = \sum_{i=1}^n a_{ii} = a_{11} + a_{22} + \dots + a_{nn}
$$

*What each symbol means:*
- **tr(A)**: Trace of matrix A
- **n**: Dimension of the square matrix
- **a_{ii}**: The diagonal entries where row index equals column index
- **Sum of diagonals**: Adding up all entries from top-left to bottom-right

### Eigenvalue Equation

$$
A\mathbf{v} = \lambda\mathbf{v}
$$

*What each symbol means:*
- **A**: An n×n square matrix representing a linear transformation
- **v**: Eigenvector (nonzero vector that doesn't change direction under transformation)
- **\lambda (lambda)**: Eigenvalue (scalar factor by which the eigenvector is stretched)
- **Av**: Result of applying matrix A to vector v
- **\lambda v**: Eigenvector scaled by eigenvalue

### Characteristic Polynomial

$$
\text{det}(A - \lambda I) = 0
$$

*What each symbol means:*
- **A**: The original square matrix
- **\lambda (lambda)**: Variable representing eigenvalues
- **I**: Identity matrix (1s on diagonal, 0s elsewhere)
- **A - \lambda I**: Matrix obtained by subtracting \lambda from each diagonal entry of A
- **\text{det}(A - \lambda I)**: Determinant of the modified matrix
- **= 0**: Setting determinant to zero gives polynomial whose roots are eigenvalues

## Worked Example by Hand

> 🧮 Using the example from the slides: Find the determinant of the matrix formed by composing vectors [2, 0, -8], [6, 1, 0], and [1, 4, -1]


### Step 1: Write down the matrix

$$
A = \begin{pmatrix}
2 & 0 & -8 \\
6 & 1 & 0 \\
1 & 4 & -1
\end{pmatrix}
$$

### Step 2: Apply the 3x3 determinant formula

$$
\text{det}(A) = 2(1 \cdot (-1) - 0 \cdot 4) - 0(6 \cdot (-1) - 0 \cdot 1) + (-8)(6 \cdot 4 - 1 \cdot 1)
$$

### Step 3: Calculate each term

- First term: $2(1 \cdot (-1) - 0 \cdot 4) = 2(-1 - 0) = 2(-1) = -2$
- Second term: $-0(6 \cdot (-1) - 0 \cdot 1) = -0(-6 - 0) = -0(-6) = 0$
- Third term: $(-8)(6 \cdot 4 - 1 \cdot 1) = (-8)(24 - 1) = (-8)(23) = -184$

### Step 4: Sum the terms

$$
\text{det}(A) = -2 + 0 + (-184) = -186
$$

### Step 5: Take absolute value (as shown in slides)

$$
|\text{det}(A)| = |-186| = 186
$$

**Answer: |det(A)| = 186**

<em>This means the transformation represented by matrix A scales volumes by a factor of 186.</em>

## Watch Out: Common Mistakes & Exam Traps

> ⚠️ **Common mistakes to avoid:**
> 1. **Confusing determinant with trace**: Determinant involves products and subtraction (ad-bc), trace is just sum of diagonals (a+d). They measure different properties!
> 2. **Forgetting the sign in 3x3 determinant**: The expansion follows + - + pattern for the first row, not + + +. Missing a minus sign gives wrong answer.
> 3. **Thinking zero determinant means zero matrix**: A zero determinant means the matrix is singular (collapses space), not that all entries are zero. Example: [[1,1],[1,1]] has det=0 but isn't the zero matrix.
> 4. **Mixing up eigenvalues and eigenvectors**: Remember: Av = λv. The vector v doesn't change direction (eigenvector), λ tells how much it stretches (eigenvalue).
> 5. **Assuming all matrices have real eigenvalues**: Rotation matrices like [[0,-1],[1,0]] have complex eigenvalues ±i. Only symmetric matrices are guaranteed real eigenvalues.

### Question 2 (Concept): If you multiply a matrix by 2, what happens to its determinant?

<details>
<summary>Click to reveal answer</summary>

For an n×n matrix, multiplying by 2 multiplies the determinant by 2ⁿ. For 2x2: det(2A) = 4·det(A); for 3x3: det(2A) = 8·det(A).

</details>

### Question 3 (Calculation): Find the determinant of $\begin{pmatrix} 3 & 1 \\ 2 & 4 \end{pmatrix}$

<details>
<summary>Click to reveal answer</summary>

det = (3×4) - (1×2) = 12 - 2 = 10

</details>

### Question 4 (Calculation): Find the determinant of $\begin{pmatrix} 1 & 0 & 2 \\ 3 & -1 & 1 \\ 2 & 1 & 0 \end{pmatrix}$

<details>
<summary>Click to reveal answer</summary>

Using first row: 1((-1)×0 - 1×1) - 0(3×0 - 1×2) + 2(3×1 - (-1)×2) = 1(0-1) - 0 + 2(3+2) = -1 + 10 = 9

</details>

### Question 5 (Explain it simply): Why do we care about eigenvectors in machine learning?

<details>
<summary>Click to reveal answer</summary>

Eigenvectors show the directions where data varies the most (in PCA) or where transformations act simply (stretching only). They reveal the "natural axes" of our data, helping us reduce dimensions while preserving important patterns.

</details>

**Key Definitions**
