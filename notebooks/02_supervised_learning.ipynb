# Lecture 02: Supervised Learning

## Classification Problems and Hypothesis Functions

## Big Idea
Supervised learning algorithms learn from labeled examples to make predictions on new, unseen data.

## Everyday Metaphor

> 💡 **Supervised Learning = learning with a teacher**: Imagine learning to recognize fruits with a teacher who shows you examples and tells you what each fruit is called. After seeing enough labeled examples, you can identify new fruits on your own.
> 
> **Labeled Training Data = flashcards with answers**: Each example consists of an input (features) and the correct output (label), like flashcards that show a picture on one side and the name on the other.
> 
> **Hypothesis Function = your internal model**: As you study the flashcards, you develop mental rules for identifying fruits. This internal model is your hypothesis function - it takes an input (what you see) and predicts an output (the fruit name).

## Why It Matters
- Supervised learning powers most practical AI applications today
- Enables computers to make accurate predictions from historical data
- Forms the foundation for image recognition, speech-to-text, and recommendation systems
- Provides a framework for understanding how machines learn from experience
- Essential for developing AI systems that assist in medical diagnosis, finance, and autonomous vehicles

## Formulas


### Hypothesis Function for Linear Regression

$$
h_\theta(x) = \theta^T x = \theta_0 + \theta_1 x_1 + \theta_2 x_2 + \dots + \theta_n x_n
$$

*What each symbol means:*
- **h_\theta(x)**: Hypothesis function predicting output for input x
- **\theta**: Parameter vector (weights) that the learning algorithm will learn
- **x**: Input feature vector (with x_0 = 1 for the intercept term)
- **\theta^T x**: Dot product of parameters and features

### Cost Function (Mean Squared Error)

$$
J(\theta) = \frac{1}{2m} \sum_{i=1}^m \left( h_\theta(x^{(i)}) - y^{(i)} \right)^2
$$

*What each symbol means:*
- **J(\theta)**: Cost function measuring prediction error
- **m**: Number of training examples
- **x^{(i)}, y^{(i)}**: i-th training example (input and label)
- **h_\theta(x^{(i)})**: Prediction for the i-th example
- **\frac{1}{2m}**: Normalization factor (the 1/2 simplifies derivative calculations)

### Gradient Descent Update Rule

$$
\theta_j := \theta_j - \alpha \frac{\partial}{\partial \theta_j} J(\theta)
$$

*What each symbol means:*
- **\theta_j**: j-th parameter being updated
- **:=\**: Assignment operator (update the parameter)
- **\alpha**: Learning rate (controls step size in parameter space)
- **\frac{\partial}{\partial \theta_j} J(\theta)**: Partial derivative of cost with respect to parameter j

### Sigmoid Function for Logistic Regression

$$
g(z) = \frac{1}{1 + e^{-z}}
$$

*What each symbol means:*
- **g(z)**: Sigmoid (logistic) function
- **z**: Input to the sigmoid function (typically z = \theta^T x)
- **e^{-z}**: Exponential function
- **\frac{1}{1 + e^{-z}}**: Output ranges from 0 to 1, perfect for probabilities

### Hypothesis Function for Logistic Regression

$$
h_\theta(x) = g(\theta^T x) = \frac{1}{1 + e^{-\theta^T x}}
$$

*What each symbol means:*
- **h_\theta(x)**: Hypothesis outputting probability of positive class
- **g(\theta^T x)**: Sigmoid applied to linear prediction
- **\theta^T x**: Linear combination of features and parameters
- **\frac{1}{1 + e^{-\theta^T x}}**: Sigmoid function as defined above

## Worked Example by Hand

> 🧮 Simple linear regression: Find the best-fit line for points (1,1), (2,3), (3,2)

### Step 1: Set up the problem

We want to find parameters \theta_0 and \theta_1 for h_\theta(x) = \theta_0 + \theta_1 x

### Step 2: Write the cost function

$$
J(\theta_0, \theta_1) = \frac{1}{2 \cdot 3} \left[ (\theta_0 + \theta_1 \cdot 1 - 1)^2 + (\theta_0 + \theta_1 \cdot 2 - 3)^2 + (\theta_0 + \theta_1 \cdot 3 - 2)^2 \right]
$$

### Step 3: Take partial derivatives

$$
\frac{\partial J}{\partial \theta_0} = \frac{1}{3} \left[ (\theta_0 + \theta_1 - 1) + (\theta_0 + 2\theta_1 - 3) + (\theta_0 + 3\theta_1 - 2) \right]
$$

$$
\frac{\partial J}{\partial \theta_1} = \frac{1}{3} \left[ (\theta_0 + \theta_1 - 1) \cdot 1 + (\theta_0 + 2\theta_1 - 3) \cdot 2 + (\theta_0 + 3\theta_1 - 2) \cdot 3 \right]
$$

### Step 4: Set derivatives to zero and solve

$$
\begin{cases}
\frac{\partial J}{\partial \theta_0} = 0 \\
\frac{\partial J}{\partial \theta_1} = 0
\end{cases}
\implies
\begin{cases}
3\theta_0 + 6\theta_1 = 6 \\
2\theta_0 + 14\theta_1 = 11
\end{cases}
\implies
\theta_0 = \frac{1}{5}, \theta_1 = \frac{4}{5}
$$

### Step 5: Write the final hypothesis

$$
h_\theta(x) = \frac{1}{5} + \frac{4}{5} x
$$

**Answer: h_\theta(x) = 0.2 + 0.4x**

<em>This line best fits the three training points in the least-squares sense.</em>

## Watch Out: Common Mistakes & Exam Traps

> ⚠️ **Common mistakes to avoid:**
> 1. **Confusing correlation with causation**: Just because features are correlated with labels doesn't mean they cause the outcome. Supervised learning finds patterns, not necessarily causal relationships.
> 2. **Overfitting to training data**: A hypothesis that perfectly fits training data may fail on new data. Always check performance on a separate validation set.
> 3. **Choosing wrong learning rate**: Too large causes divergence, too small makes learning impractically slow. Always try multiple learning rates.
> 4. **Forgetting feature scaling**: Gradient descent converges much faster when features are on similar scales. Standardization or normalization is often essential.
> 5. **Misinterpreting logistic regression outputs**: The output is a probability, not a binary classification. You need a threshold (usually 0.5) to convert to class predictions.

<details>

The 1/2 factor simplifies the derivative calculation during gradient descent. The 1/m factor averages the error over all training examples.

## Linear Regression Implementation

> 📝 Implement linear regression from scratch using numpy and compare with scikit-learn.

First, let's import the necessary libraries:

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Our training data from the worked example
X = np.array([[1], [2], [3]])  # Features
Y = np.array([1, 3, 2])        # Labels

# Add intercept term (column of ones)
X_b = np.c_[np.ones((3, 1)), X]  # Now X_b has shape (3, 2)

# Batch Gradient Descent implementation
def gradient_descent(X, Y, learning_rate=0.1, n_iterations=1000):
    m = len(Y)  # number of training examples
    theta = np.zeros(2)  # initialize parameters to zero
    
    # Store history for plotting
    theta_history = []
    cost_history = []
    
    for iteration in range(n_iterations):
        # Predictions
        predictions = X_b.dot(theta)
        
        # Cost (MSE)
        cost = np.sum((predictions - Y) ** 2) / (2 * m)
        
        # Gradients
        gradients = X_b.T.dot(predictions - Y) / m
        
        # Update parameters
        theta = theta - learning_rate * gradients
        
        # Store history
        theta_history.append(theta.copy())
        cost_history.append(cost)
    
    return theta, theta_history, cost_history

# Run gradient descent
theta, theta_history, cost_history = gradient_descent(X, Y, learning_rate=0.1)

print("Parameters from gradient descent:")
print(f"\theta_0 = {theta[0]:.4f}")
print(f"\theta_1 = {theta[1]:.4f}")

# Compare with scikit-learn
sklearn_model = LinearRegression()
sklearn_model.fit(X, Y)

print("\nParameters from scikit-learn:")
print(f"\theta_0 (intercept) = {sklearn_model.intercept_:.4f}")
print(f"\theta_1 (coefficient) = {sklearn_model.coef_[0]:.4f}")

# Make predictions
X_test = np.array([[0], [1], [2], [3], [4]])
X_test_b = np.c_[np.ones((5, 1)), X_test]

y_predict = X_test_b.dot(theta)
y_predict_sklearn = sklearn_model.predict(X_test)

print("\nPredictions:")
print("x\tGD Prediction\tSKLearn Prediction")
for i in range(len(X_test)):
    print(f"{X_test[i][0]}\t{y_predict[i]:.4f}\t\t{y_predict_sklearn[i]:.4f}")

# Plot the results
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.scatter(X, Y, color='red', s=100, zorder=5)
plt.plot(X_test, y_predict, 'b-', linewidth=2, label='Gradient Descent')
plt.plot(X_test, y_predict_sklearn, 'g--', linewidth=2, label='Scikit-learn')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Linear Regression Fit')
plt.legend()
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(cost_history)
plt.xlabel('Iteration')
plt.ylabel('Cost')
plt.title('Cost Function during Gradient Descent')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

## Logistic Regression Implementation

> 📝 Implement logistic regression from scratch using numpy.

import numpy as np
import matplotlib.pyplot as plt

# Create a simple dataset for binary classification
np.random.seed(42)
X1 = np.random.multivariate_normal([0, 0], [[1, 0.5], [0.5, 1]], 50)
X2 = np.random.multivariate_normal([2, 2], [[1, -0.3], [-0.3, 1]], 50)
X = np.vstack((X1, X2))
y = np.hstack((np.zeros(50), np.ones(50)))

# Add intercept term
X_b = np.c_[np.ones((100, 1)), X]

# Sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-np.clip(z, -250, 250)))  # Clip to prevent overflow

# Cost function for logistic regression
def compute_cost(X, y, theta):
    m = len(y)
    h = sigmoid(X.dot(theta))
    # Add small epsilon to prevent log(0)
    epsilon = 1e-15
    h = np.clip(h, epsilon, 1 - epsilon)
    cost = (-1/m) * np.sum(y * np.log(h) + (1 - y) * np.log(1 - h))
    return cost

# Gradient descent for logistic regression
def logistic_gradient_descent(X, y, learning_rate=0.1, n_iterations=1000):
    m = len(y)
    theta = np.zeros(X.shape[1])  # initialize parameters
    
    cost_history = []
    
    for iteration in range(n_iterations):
        z = X.dot(theta)
        h = sigmoid(z)
        
        # Cost
        cost = compute_cost(X, y, theta)
        
        # Gradient
        gradient = X.T.dot(h - y) / m
        
        # Update parameters
        theta = theta - learning_rate * gradient
        
        # Store history
        cost_history.append(cost)
    
    return theta, cost_history

# Train logistic regression model
theta, cost_history = logistic_gradient_descent(X_b, y, learning_rate=0.1)

print("Parameters from logistic regression:")
print(f"\theta_0 = {theta[0]:.4f}")
print(f"\theta_1 = {theta[1]:.4f}")
print(f"\theta_2 = {theta[2]:.4f}")

# Make predictions
probabilities = sigmoid(X_b.dot(theta))
predictions = (probabilities >= 0.5).astype(int)

accuracy = np.mean(predictions == y)
print(f"\nAccuracy: {accuracy:.4f}")

# Plot the decision boundary
plt.figure(figsize=(8, 6))

# Plot points
plt.scatter(X[y==0][:, 0], X[y==0][:, 1], color='red', label='Class 0', alpha=0.7)
plt.scatter(X[y==1][:, 0], X[y==1][:, 1], color='blue', label='Class 1', alpha=0.7)

# Plot decision boundary
x1_min, x1_max = X[:, 0].min() - 1, X[:, 0].max() + 1
x2_min, x2_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx1, xx2 = np.meshgrid(np.linspace(x1_min, x1_max, 100),
                       np.linspace(x2_min, x2_max, 100))
grid = np.c_[np.ones((xx1.ravel().size, 1)), xx1.ravel(), xx2.ravel()]
probs = sigmoid(grid.dot(theta)).reshape(xx1.shape)
plt.contour(xx1, xx2, probs, levels=[0.5], linewidths=2, colors='green')

plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Logistic Regression Decision Boundary')
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
